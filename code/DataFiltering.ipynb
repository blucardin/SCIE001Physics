```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time
```

```python
%matplotlib qt

# read the CSV file into a dataframe (i.e. like a Python spreadsheet)
df = pd.read_csv("run4-610-20251009-023613.csv")

# convert dataframe into a 2D Numpy array
a = df.to_numpy()

# the first column are the counters
counter_array = a[:,0]

# the second column are the time stamps
computer_time_stamp_array = a[:,2]

# the second column are the time stamps
arduino_time_stamp_array = a[:,2]

# the third column are your measurements: either pulseTime or distance depending on how you modified the code above
measurement_array = a[:,3]
```

```python
#to double-check everything was imported ok, plot the data again
%matplotlib qt

plt.figure()
plt.plot(arduino_time_stamp_array, measurement_array,".")  # plot data as points
plt.xlabel("Time Stamp (s)")
plt.ylabel("pulseTime (μs)")
plt.ylim(0, 1.1*max(measurement_array));   # set y axis range
```

```python
%matplotlib inline  

times = np.logical_and(arduino_time_stamp_array > 1.73 * 10 ** 8,  arduino_time_stamp_array < 1.736 * 10 ** 8)


x  = arduino_time_stamp_array[times]
y = measurement_array[times]

# 2. Plot the data

#to double-check everything was imported ok, plot the data again
plt.figure()
plt.plot(x, y, ".")
plt.xlabel("Time Stamp (s)")
plt.ylabel("pulseTime (μs)")
# plt.ylim(0, 1.1*max(measurement_array));   # set y axis range
```

```python

timestr = time.strftime("%Y%m%d-%H%M%S")        # get current date and time into a string
FILE_NAME = 'Yanni-CopyDONOTUSE-'+timestr+'.csv'     # define csv file name based on current data and time, so you don't lose track of when you took your data

df = pd.DataFrame()
df['Computer Time (microseconds)'] = computer_time_stamp_array[times]
df['Arduino Time (seconds)'] = x
df['pulseTime (microseconds)'] = y

FILE_NAME = "Good Data/" + FILE_NAME

df.to_csv(FILE_NAME)  # export that dataframe to a CSV file
```
