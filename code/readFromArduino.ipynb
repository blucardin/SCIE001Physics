```python
import serial
```

```python
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
ser = serial.Serial('/dev/cu.usbmodem14301', 115200, timeout=1)
```

```python

# prepare our CSV file that will write our data to
timestr = time.strftime("%Y%m%d-%H%M%S")        # get current date and time into a string
FILE_NAME = 'run5-20-'+timestr+'.csv'     # define csv file name based on current data and time, so you don't lose track of when you took your data

computer_time = []
arduino_time = []
pulse_time = []

try: 
    print("Reading")
    while True:
        if ser.in_waiting > 0: # Check if there's data in the receive buffer
            line = ser.readline().decode('utf-8').rstrip().split(",")
            # print(line)
            a, b, current_arduino_time, current_pulse_time = line
            assert a == "5000"
            assert b == "0"

            computer_time.append(time.time_ns() / 1000)
            arduino_time.append(float(current_arduino_time))
            pulse_time.append(float(current_pulse_time))
            # print(line)

# KeyboardInterrupt means Python is waiting for you to manually interrupt the program (using the "stop" button) or by pressing "i" twice in quick succession
except KeyboardInterrupt:

    # save your data to a CSV file
    # df = pd.DataFrame(data)      # first convert our Numpy array into a Pandas dataframe (like a spreadsheet)
    plt.figure()
    plt.plot(np.array(arduino_time), np.array(pulse_time), ".")  # plot data as points
    
    # plt.plot(np.array(arduino_time), np.array(pulse_time), ".")  # plot data as points

    plt.xlabel("Time Stamp (μs)")
    plt.ylabel("pulseTime (μs)")
    
    # 4. Display the plot
    plt.show()    
    # board.close()   # close board connection
```

```python
# Code to save, it if is good
df = pd.DataFrame()
df['Computer Time (microseconds)'] = computer_time
df['Arduino Time (seconds)'] = arduino_time
df['pulseTime (microseconds)'] = pulse_time

df.to_csv(FILE_NAME)  # export that dataframe to a CSV file
```
